Task 18: Potential Machine Learning Problems
1. Predict Late Order Delivery
Problem Type: Classification
Target: Late / Not Late
Description: Predict whether an order will be delivered later than the estimated delivery date.
2. Predict Customer Review Score
Problem Type: Regression (or Classification)
Target: review_score
Description: Predict the review score (1–5) based on order, product, and delivery details.
3. Predict Payment Value
Problem Type: Regression
Target: payment_value
Description: Predict the total payment amount for an order.
4. Customer Segmentation
Problem Type: Clustering
Target: No explicit target
Description: Group customers based on purchasing behavior, location, or spending patterns.
5. Product Recommendation
Problem Type: Recommendation System
Target: Product Category
Description: Recommend products or categories to customers based on previous purchases.
